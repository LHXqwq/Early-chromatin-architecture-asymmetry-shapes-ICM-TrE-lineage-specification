```python
import pandas as pd
import glob
import matplotlib.pyplot as plt

# =========================
# 1. 读取 gene BED
# =========================
gene_bed = pd.read_csv(
    "Sus_scrofa_gene.110.bed",
    sep="\t",
    header=None,
    names=["chr", "start", "end", "gene"]
)

# =========================
# 2. 读取 region BED（修正：保留所有列）
# =========================
def read_bed(file):
    df = pd.read_csv(file, sep="\t", header=None)
    
    # 自动判断列数
    if df.shape[1] == 3:
        df.columns = ["chr", "start", "end"]
    else:
        df = df.iloc[:, :3]
        df.columns = ["chr", "start", "end"]
    
    return df

regions = {
    "boundary": read_bed("dedoc2_IVF_2C_debin.bed"),
    "flank_80k": read_bed("dedoc2_IVF_2C_debin_flanking_80k.bed"),
    "flank_160k": read_bed("dedoc2_IVF_2C_debin_flanking_160k.bed"),
}

# =========================
# 3. 统一 chr 格式（非常关键）
# =========================
def normalize_chr(df):
    df["chr"] = df["chr"].astype(str).str.replace("chr", "")
    return df

gene_bed = normalize_chr(gene_bed)
for k in regions:
    regions[k] = normalize_chr(regions[k])

# =========================
# 4. TSS-based overlap（loop-level）
# =========================
def overlap_tss(df, region_df):
    df = df.copy()
    df["in_region"] = False

    region_dict = {
        chr_: region_df[region_df["chr"] == chr_]
        for chr_ in region_df["chr"].unique()
    }

    for i, row in df.iterrows():
        chr_, s = row["chr"], row["start"]

        if pd.isna(chr_) or chr_ not in region_dict:
            continue

        sub = region_dict[chr_]

        # TSS point-in-interval
        hit = ((sub["start"] <= s) & (sub["end"] >= s)).any()

        if hit:
            df.at[i, "in_region"] = True

    return df

# =========================
# 5. 主分析
# =========================
files = glob.glob("*_each_diffEPI_count.txt")

for f in files:
    print(f"\nProcessing {f}")

    df = pd.read_csv(f, sep="\t")

    # diff loop
    df["diff"] = (df["C1"] - df["C2"]).abs()
    diff_df = df[df["diff"] > 2]

    # merge gene坐标 → loop-level
    merged = diff_df.merge(gene_bed, on="gene", how="left")
    merged = merged.dropna(subset=["chr"])

    total = len(merged)
    print("Total diff loops:", total)

    stats = {}

    for name, region in regions.items():
        g = overlap_tss(merged, region)

        count = g["in_region"].sum()
        stats[name] = count

        print(f"{name}: {count}/{total} = {count/total:.4f}")

    # =========================
    # 6. 饼图（boundary）
    # =========================
    inside = stats["boundary"]
    outside = total - inside

    plt.figure()
    plt.pie(
        [inside, outside],
        labels=["In boundary", "Outside"],
        autopct="%1.1f%%"
    )
    plt.title(f"{f} (loop-level TSS)")
    plt.savefig(f"{f}_boundary_pie.png", dpi=300)
    plt.close()

print("\nDone.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import mannwhitneyu
from IPython.display import SVG, display

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# =========================
# 1. 读取数据
# =========================
df = pd.read_csv("显著差异边界频率比较.txt", sep="\t")

# =========================
# 2. ratio（max/min，忽略0）
# =========================
def compute_ratio(row):
    h1 = row["H1"]
    h2 = row["H2"]

    if h1 == 0 or h2 == 0:
        return np.nan

    return max(h1, h2) / min(h1, h2)

df["ratio"] = df.apply(compute_ratio, axis=1)
df = df.dropna(subset=["ratio"])

# =========================
# 3. 固定顺序
# =========================
order = ["IVF-2C", "IVF-4C", "PA-2C", "PA-4C"]

data = [df[df["type"] == t]["ratio"].values for t in order]

# =========================
# 4. 统计检验
# =========================
ivf_2c = df[df["type"] == "IVF-2C"]["ratio"]
ivf_4c = df[df["type"] == "IVF-4C"]["ratio"]

pa_2c  = df[df["type"] == "PA-2C"]["ratio"]
pa_4c  = df[df["type"] == "PA-4C"]["ratio"]

p_ivf = mannwhitneyu(ivf_2c, ivf_4c, alternative="two-sided").pvalue
p_pa  = mannwhitneyu(pa_2c, pa_4c,  alternative="two-sided").pvalue

# =========================
# 5. 画图
# =========================
plt.figure(figsize=(7, 4))

plt.boxplot(data, labels=order, showfliers=False)

plt.ylabel("H1/H2 ratio (max/min)")
plt.title("Boundary frequency asymmetry")

plt.xticks(rotation=30)

# =========================
# 6. 标注显著性
# =========================
def add_sig(x1, x2, y, p):
    plt.plot([x1, x1, x2, x2], [y, y+0.1, y+0.1, y], lw=1)

    if p < 1e-4:
        text = "****"
    elif p < 1e-3:
        text = "***"
    elif p < 1e-2:
        text = "**"
    elif p < 0.05:
        text = "*"
    else:
        text = "ns"

    plt.text((x1+x2)/2, y+0.12, text, ha='center')

y_max = max(df["ratio"])

# IVF (1 vs 2)
add_sig(1, 2, y_max*1.05, p_ivf)

# PA (3 vs 4)
add_sig(3, 4, y_max*1.25, p_pa)

plt.ylim(0, y_max*1.5)

plt.tight_layout()
plt.savefig('Boundary_frequency_asymmetry.svg', dpi=300, bbox_inches='tight', transparent=True)
plt.show()

# =========================
# 7. 输出 p-value
# =========================
print("IVF 2C vs 4C p =", p_ivf)
print("PA  2C vs 4C p =", p_pa)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib_venn import venn2
from IPython.display import SVG, display

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

files = {
    "Morula Boundary": "IVF_morula_boundaries_sum_pvalue.txt",
    "Morula Boundary ± 80k": "IVF_morula_boundaries_80k_sum_pvalue.txt",
    "Morula Boundary ± 160k": "IVF_morula_boundaries_160k_sum_pvalue.txt"
}

fig, axes = plt.subplots(3, 4, figsize=(16, 12))

for i, (res, f) in enumerate(files.items()):

    df = pd.read_csv(f, sep="\t")

    df["bin_id"] = df["chr"].astype(str) + ":" + df["start"].astype(str) + "-" + df["end"].astype(str)

    # =========================
    # 2C
    # =========================
    df_2c = df[df["p_value_2c"] < 0.05]

    set_2c_gt = set(df_2c.loc[df_2c["IVF-2C-H1"] > df_2c["IVF-2C-H2"], "bin_id"])
    set_2c_lt = set(df_2c.loc[df_2c["IVF-2C-H1"] < df_2c["IVF-2C-H2"], "bin_id"])

    # =========================
    # 4C
    # =========================
    df_4c = df[df["p_value_4c"] < 0.05]

    set_4c_gt = set(df_4c.loc[df_4c["IVF-4C-H1"] > df_4c["IVF-4C-H2"], "bin_id"])
    set_4c_lt = set(df_4c.loc[df_4c["IVF-4C-H1"] < df_4c["IVF-4C-H2"], "bin_id"])

    # =========================
    # 1) same-direction
    # =========================
    ax1 = axes[i, 0]
    venn2([set_2c_gt, set_4c_gt], set_labels=("2C", "4C"), ax=ax1)
    ax1.set_title(f"{res}\n H1 > H2")

    ax2 = axes[i, 1]
    venn2([set_2c_lt, set_4c_lt], set_labels=("2C", "4C"), ax=ax2)
    ax2.set_title(f"{res}\n H1 < H2")

    # =========================
    # 2) SWITCH analysis
    # =========================

    # 2C H1>H2  vs 4C H2>H1
    ax3 = axes[i, 2]
    venn2([set_2c_gt, set_4c_lt], set_labels=("2C H1>H2", "4C H2>H1"), ax=ax3)
    ax3.set_title(f"{res}\n H1 switch → H2")

    # 2C H2>H1 vs 4C H1>H2
    ax4 = axes[i, 3]
    venn2([set_2c_lt, set_4c_gt], set_labels=("2C H2>H1", "4C H1>H2"), ax=ax4)
    ax4.set_title(f"{res}\n H2 switch → H1")

plt.tight_layout()
plt.savefig("IVF_boundary_switch_overlap_12panel.svg", dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy.stats import mannwhitneyu
from IPython.display import SVG, display

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# =========================
# 1. order
# =========================
order = [
    "2C:H1>H2",
    "2C:H1<H2",
    "4C:H1>H2",
    "4C:H1<H2"
]

# =========================
# 2. files
# =========================
file_groups = {
    "Morula Boundary": "IVF_morula_boundaries_sum_pvalue.txt",
    "Morula Boundary ± 80k": "IVF_morula_boundaries_80k_sum_pvalue.txt",
    "Morula Boundary ± 160k": "IVF_morula_boundaries_160k_sum_pvalue.txt"
}

# =========================
# 3. 颜色定义（核心）
# =========================
def get_colors(group):
    if group.startswith("2C"):
        return "#A73BDD", "#E7A31B"   # H1, H2
    else:
        return "#2292A4", "#E22720"

# =========================
# 4. build dataframe
# =========================
all_plot = []

for res, f in file_groups.items():

    df = pd.read_csv(f, sep="\t")

    df.columns = [
        "chr", "start", "end",
        "H1_2C", "H2_2C",
        "H1_4C", "H2_4C",
        "p2c", "p4c"
    ]

    for stage, pcol, h1col, h2col in [
        ("2C", "p2c", "H1_2C", "H2_2C"),
        ("4C", "p4c", "H1_4C", "H2_4C")
    ]:

        df_s = df[df[pcol] < 0.05]

        for mask, name in [
            (df_s[h1col] > df_s[h2col], f"{stage}:H1>H2"),
            (df_s[h1col] < df_s[h2col], f"{stage}:H1<H2")
        ]:

            sub = df_s[mask]

            h1 = sub[h1col].values
            h2 = sub[h2col].values

            h1 = h1[~np.isnan(h1)]
            h2 = h2[~np.isnan(h2)]

            all_plot.append(pd.DataFrame({
                "value": h1,
                "group": name,
                "cell": "H1",
                "res": res
            }))

            all_plot.append(pd.DataFrame({
                "value": h2,
                "group": name,
                "cell": "H2",
                "res": res
            }))

plot_df = pd.concat(all_plot, ignore_index=True)
plot_df["group"] = pd.Categorical(plot_df["group"], categories=order, ordered=True)

# =========================
# 5. 显著性函数
# =========================
def get_p_annotation(p):
    if p < 1e-4:
        return "****"
    elif p < 1e-3:
        return "***"
    elif p < 1e-2:
        return "**"
    elif p < 0.05:
        return "*"
    else:
        return "ns"

# =========================
# 6. plot
# =========================
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

for i, res in enumerate(file_groups.keys()):

    ax = axes[i]
    tmp = plot_df[plot_df["res"] == res]

    # 先画空白 violin（占位）
    sns.violinplot(
        data=tmp,
        x="group",
        y="value",
        hue="cell",
        split=True,
        inner="quartile",
        palette={"H1": "white", "H2": "white"},
        ax=ax
    )

    ax.set_title(res)
    ax.set_xlabel("")
    ax.tick_params(axis='x', rotation=45)
    ax.set_ylabel("Cell Number" if i == 0 else "")

    polys = ax.collections

    y_max = tmp["value"].max()
    y_text = y_max * 1.05

    for gi, group in enumerate(order):

        # 数据
        h1 = tmp[(tmp["group"] == group) & (tmp["cell"] == "H1")]["value"].values
        h2 = tmp[(tmp["group"] == group) & (tmp["cell"] == "H2")]["value"].values

        if len(h1) > 2 and len(h2) > 2:
            _, p = mannwhitneyu(h1, h2)
        else:
            p = 1.0

        sig = get_p_annotation(p)

        # 取颜色
        c_h1, c_h2 = get_colors(group)

        # 根据方向决定谁“主导颜色”
        if "H1>H2" in group:
            left_color = c_h1
            right_color = c_h2
        else:
            left_color = c_h1
            right_color = c_h2

        # 小提琴对象
        h1_poly = polys[gi * 2]
        h2_poly = polys[gi * 2 + 1]

        # 上色（关键）
        h1_poly.set_facecolor(left_color)
        h2_poly.set_facecolor(right_color)

        for p_obj in [h1_poly, h2_poly]:
            p_obj.set_edgecolor("black")
            p_obj.set_alpha(0.85)

        # 显著性
        ax.text(gi, y_text, sig, ha='center', fontsize=12, color="red")

    # 去掉每个子图 legend
    if ax.get_legend():
        ax.get_legend().remove()

    # 只在最后一个放 legend
    if i == 2:
        legend_handles = [
            mpatches.Patch(color="#A73BDD", label="IVF-2C H1"),
            mpatches.Patch(color="#E7A31B", label="IVF-2C H2"),
            mpatches.Patch(color="#2292A4", label="IVF-4C H1"),
            mpatches.Patch(color="#E22720", label="IVF-4C H2"),
        ]

        ax.legend(
            handles=legend_handles,
            loc="center left",
            bbox_to_anchor=(1.02, 0.5),
            frameon=False
        )

plt.tight_layout()
plt.savefig("IVF_boundary_violin_3panel.svg", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy.stats import mannwhitneyu
from IPython.display import SVG, display

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# =========================
# 1. order
# =========================
order = [
    "2C:H1>H2",
    "2C:H1<H2",
    "4C:H1>H2",
    "4C:H1<H2"
]

# =========================
# 2. files
# =========================
file_groups = {
    "Morula Boundary": "PA_morula_boundaries_sum_pvalue.txt",
    "Morula Boundary ± 80k": "PA_morula_boundaries_80k_sum_pvalue.txt",
    "Morula Boundary ± 160k": "PA_morula_boundaries_160k_sum_pvalue.txt"
}

# =========================
# 3. 颜色定义（核心）
# =========================
def get_colors(group):
    if group.startswith("2C"):
        return "#7dace4", "#ff165d"   # H1, H2
    else:
        return "green", "orange"


# =========================
# 4. build dataframe
# =========================
all_plot = []

for res, f in file_groups.items():

    df = pd.read_csv(f, sep="\t")

    df.columns = [
        "chr", "start", "end",
        "H1_2C", "H2_2C",
        "H1_4C", "H2_4C",
        "p2c", "p4c"
    ]

    for stage, pcol, h1col, h2col in [
        ("2C", "p2c", "H1_2C", "H2_2C"),
        ("4C", "p4c", "H1_4C", "H2_4C")
    ]:

        df_s = df[df[pcol] < 0.05]

        for mask, name in [
            (df_s[h1col] > df_s[h2col], f"{stage}:H1>H2"),
            (df_s[h1col] < df_s[h2col], f"{stage}:H1<H2")
        ]:

            sub = df_s[mask]

            h1 = sub[h1col].values
            h2 = sub[h2col].values

            h1 = h1[~np.isnan(h1)]
            h2 = h2[~np.isnan(h2)]

            all_plot.append(pd.DataFrame({
                "value": h1,
                "group": name,
                "cell": "H1",
                "res": res
            }))

            all_plot.append(pd.DataFrame({
                "value": h2,
                "group": name,
                "cell": "H2",
                "res": res
            }))

plot_df = pd.concat(all_plot, ignore_index=True)
plot_df["group"] = pd.Categorical(plot_df["group"], categories=order, ordered=True)

# =========================
# 5. 显著性函数
# =========================
def get_p_annotation(p):
    if p < 1e-4:
        return "****"
    elif p < 1e-3:
        return "***"
    elif p < 1e-2:
        return "**"
    elif p < 0.05:
        return "*"
    else:
        return "ns"

# =========================
# 6. plot
# =========================
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

for i, res in enumerate(file_groups.keys()):

    ax = axes[i]
    tmp = plot_df[plot_df["res"] == res]

    # 先画空白 violin（占位）
    sns.violinplot(
        data=tmp,
        x="group",
        y="value",
        hue="cell",
        split=True,
        inner="quartile",
        palette={"H1": "white", "H2": "white"},
        ax=ax
    )

    ax.set_title(res)
    ax.set_xlabel("")
    ax.tick_params(axis='x', rotation=45)
    ax.set_ylabel("Cell Number" if i == 0 else "")

    polys = ax.collections

    y_max = tmp["value"].max()
    y_text = y_max * 1.05

    for gi, group in enumerate(order):

        # 数据
        h1 = tmp[(tmp["group"] == group) & (tmp["cell"] == "H1")]["value"].values
        h2 = tmp[(tmp["group"] == group) & (tmp["cell"] == "H2")]["value"].values

        if len(h1) > 2 and len(h2) > 2:
            _, p = mannwhitneyu(h1, h2)
        else:
            p = 1.0

        sig = get_p_annotation(p)

        # 取颜色
        c_h1, c_h2 = get_colors(group)

        # 根据方向决定谁“主导颜色”
        if "H1>H2" in group:
            left_color = c_h1
            right_color = c_h2
        else:
            left_color = c_h1
            right_color = c_h2

        # 小提琴对象
        h1_poly = polys[gi * 2]
        h2_poly = polys[gi * 2 + 1]

        # 上色（关键）
        h1_poly.set_facecolor(left_color)
        h2_poly.set_facecolor(right_color)

        for p_obj in [h1_poly, h2_poly]:
            p_obj.set_edgecolor("black")
            p_obj.set_alpha(0.85)

        # 显著性
        ax.text(gi, y_text, sig, ha='center', fontsize=12, color="red")

    # 去掉每个子图 legend
    if ax.get_legend():
        ax.get_legend().remove()

    # 只在最后一个放 legend
    if i == 2:
        legend_handles = [
            mpatches.Patch(color="#A73BDD", label="PA-2C H1"),
            mpatches.Patch(color="#E7A31B", label="PA-2C H2"),
            mpatches.Patch(color="#2292A4", label="PA-4C H1"),
            mpatches.Patch(color="#E22720", label="PA-4C H2"),
        ]

        ax.legend(
            handles=legend_handles,
            loc="center left",
            bbox_to_anchor=(1.02, 0.5),
            frameon=False
        )

plt.tight_layout()
plt.savefig("PA_boundary_violin_3panel.svg", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import glob

# =========================
# 1. 读取 gene mapping
# =========================
map_df = pd.read_csv(
    "Sus_scrofa.Sscrofa11.1.110.geneid.txt",
    sep="\t"
)

gene_map = dict(zip(map_df["genes"], map_df["gene_name"]))

# =========================
# 2. 批量处理所有 geneid 文件
# =========================
files = glob.glob("*geneid.txt")

for f in files:

    if f == "Sus_scrofa.Sscrofa11.1.110.geneid.txt":
        continue

    df = pd.read_csv(f, header=None)

    df["gene_name"] = df[0].map(gene_map)

    # 如果没有match，保留原ID
    df["gene_name"] = df["gene_name"].fillna(df[0])

    out = f.replace("_geneid.txt", "_genename.txt")

    df["gene_name"].to_csv(out, index=False, header=False)

    print(f"done: {f} -> {out}")
```
