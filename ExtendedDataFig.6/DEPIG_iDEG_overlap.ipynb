```python
#IVF 2C diff e-p genes，inter-blastomere DEGs
from matplotlib_venn import venn2, venn2_circles
import matplotlib.pyplot as plt
import matplotlib
from scipy.stats import hypergeom
from IPython.display import SVG, display

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# 重要提示：需要根据实际研究背景修改总基因数M
M = 20289  # 假设基因组总基因数（需按实际调整！）
subset_values = (5552, 510, 187)  # (A_only, B_only, AB)

# ====================
# 统计检验计算
# ====================
# 计算Jaccard相似性系数
union = subset_values[0] + subset_values[1] + subset_values[2]
jaccard = subset_values[2] / union

# 超几何检验计算
n_A = subset_values[0] + subset_values[2]  # 2C基因总数
n_B = subset_values[1] + subset_values[2]  # 4C基因总数
k = subset_values[2]  # 重叠基因数

# 超几何分布公式：P(X >= k)
p_value = hypergeom.sf(
    k-1,  # 生存函数计算的是P(X > k-1)即P(X >=k)
    M,    # 总基因数
    n_A,  # 成功元素数（2C基因）
    n_B   # 抽样数（4C基因）
)

# ====================
# 可视化参数设置
# ====================
matplotlib.rcParams.update({
    'font.size': 10,
    'font.weight': 'bold'
})
plt.rcParams['font.family'] = 'DejaVu Sans'  # Linux通用字体
# ====================
# 绘制专业图表
# ====================
fig, ax = plt.subplots()

# 绘制Venn图主体
venn = venn2(
    subsets=subset_values,
    set_labels=(
        f'IVF 2C DEPIGs', 
        f'IVF 2C iDEGs'
    ),
    set_colors=('#1f77b4', '#ff7f0e'),
    alpha=0.7,
    ax=ax
)

print(f"Hypergeometric test: p = {p_value:.1e}")
print(f"Jaccard index = {jaccard:.3f}")
      

# ====================
# 图形增强
# ====================
# 添加虚线轮廓
venn2_circles(
    subsets=subset_values, 
    linestyle='--', 
    linewidth=1, 
    ax=ax,
    color='gray'
)

# ====================
# 输出控制
# ====================
plt.tight_layout()
plt.savefig('IVF_2C_DEPIGs-iDEGs_overlap.svg', dpi=300, bbox_inches='tight', transparent=True)
#display(SVG(filename='IVF_2C_EPI_inter-blastomere_overlap.svg'))
```
