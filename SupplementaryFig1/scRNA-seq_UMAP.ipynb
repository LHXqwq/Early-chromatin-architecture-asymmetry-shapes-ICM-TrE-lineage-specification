```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
from IPython.display import SVG, display

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# 自定义颜色（根据你的数据修改或扩展）
color_map = {
    "IVF-Zygote": "#ffe119",  # 黄色
    "IVF-2C": "#3cb44b",      # 绿色
    "IVF-4C": "#e6194b"       # 红色
}

# 读取 IVF_118cells_UMAP_coordinates.txt
umap_df = pd.read_csv("IVF_118cells_UMAP_coordinates.txt", sep="\t")


# 绘图参数
ds = 10
fig, ax = plt.subplots(figsize=(6, 4))
ax.set_xlabel('UMAP-1', fontsize=20)
ax.set_ylabel('UMAP-2', fontsize=20)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.tick_params(axis='both', which='both', length=0)
ax.set_xticklabels([])
ax.set_yticklabels([])
ax.set_title('UMAP by Genes expression in IVF embryos', fontsize=18)

# 绘制散点图
for cell_type, color in color_map.items():
    sub_df = umap_df[umap_df['Cell_Class'] == cell_type]
    ax.scatter(
        sub_df['UMAP_1'],
        sub_df['UMAP_2'],
        c=color,
        s=ds,
        edgecolors='none',
        alpha=0.8,
        label=cell_type,
        rasterized=True
    )

# 图例
ax.legend(
    markerscale=5,
    prop={'size': 10},
    bbox_to_anchor=(1.05, 1),
    loc='upper left',
    fontsize=15,
    frameon=False
)

plt.tight_layout()

# 保存为 PDF

plt.savefig('IVF_118cells_umap_plot.svg', dpi=300, bbox_inches='tight', transparent=True)
```

```python
#PA_IVF_198cells_UMAP_coordinates
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.lines as mlines

# 字体 & SVG 设置
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# 读取数据
umap_df = pd.read_csv("PA_IVF_198cells_UMAP_coordinates.csv", sep=",")

umap_df.loc[(umap_df['source']=='PA') & (umap_df['stage']=='Zygote'), 'stage'] = 'MII'

# 颜色映射（stage对应颜色），MII和Zygote用同色
color_map = {
    "Zygote": "#ffe119",
    "MII": "#ffe119",   # MII用同Zygote色
    "2C": "#3cb44b",
    "4C": "#e6194b",
    "Other": "#a9a9a9"
}

# 形状映射（source对应形状）
shape_map = {
    "IVF": "^",  # 三角形
    "PA": "o",   # 圆形
    "Other": "x"
}

point_size = 15
alpha_val = 0.8
figsize = (6, 4)
label_fontsize = 20
legend_fontsize = 12

fig, ax = plt.subplots(figsize=figsize)
ax.set_xlabel('UMAP-1', fontsize=label_fontsize)
ax.set_ylabel('UMAP-2', fontsize=label_fontsize)
ax.set_title('UMAP of PA & IVF embryos', fontsize=label_fontsize)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='both', which='both', length=0)
ax.set_xticklabels([])
ax.set_yticklabels([])

# 绘制散点，颜色用stage对应颜色，形状用source对应形状
for source in umap_df['source'].unique():
    for stage in umap_df['stage'].unique():
        sub_df = umap_df[(umap_df['source'] == source) & (umap_df['stage'] == stage)]
        if not sub_df.empty:
            # 用stage本身作为颜色键，如果是MII，颜色映射表中有定义，自动用黄色
            color_key = stage
            marker_shape = shape_map.get(source, 'x')
            if source == "IVF":
                ax.scatter(
                    sub_df['UMAP_1'], sub_df['UMAP_2'],
                    c=color_map.get(color_key, "#000000"),
                    marker=marker_shape,
                    s=point_size,
                    alpha=alpha_val,
                    label=f"{source}-{stage}",
                    edgecolors='none',
                    rasterized=True
                )
            elif source == "PA":
                ax.scatter(
                    sub_df['UMAP_1'], sub_df['UMAP_2'],
                    facecolors='none',
                    edgecolors=color_map.get(color_key, "#000000"),
                    marker=marker_shape,
                    s=point_size,
                    alpha=alpha_val,
                    label=f"{source}-{stage}",
                    rasterized=True
                )
            else:
                ax.scatter(
                    sub_df['UMAP_1'], sub_df['UMAP_2'],
                    c=color_map.get(color_key, "#000000"),
                    marker=marker_shape,
                    s=point_size,
                    alpha=alpha_val,
                    label=f"{source}-{stage}",
                    rasterized=True
                )

# 定义明确顺序
source_order = ['PA', 'IVF']
stage_order = {
    'PA': ['MII', '2C', '4C'],
    'IVF': ['Zygote', '2C', '4C']
}

# 绘图（不变，保持你已有代码）

# 去重并排序图例句柄
handles, labels = ax.get_legend_handles_labels()
label_to_handle = dict(zip(labels, handles))

new_labels = []
for source in source_order:
    for stage in stage_order[source]:
        label = f"{source}-{stage}"
        if label in label_to_handle:
            new_labels.append(label)

new_handles = [label_to_handle[label] for label in new_labels]

ax.legend(
    new_handles,
    new_labels,
    title="Source-Stage",
    title_fontsize=legend_fontsize,
    markerscale=1.5,
    prop={'size': legend_fontsize},
    bbox_to_anchor=(1.05, 1),
    loc='upper left',
    frameon=False
)


plt.tight_layout()
plt.savefig('PA_IVF_198cells_umap_plot.svg', dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```
