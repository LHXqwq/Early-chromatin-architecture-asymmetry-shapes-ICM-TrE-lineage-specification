```python
#IVF 去掉 极端百分位（如 2.5% 和 97.5% 之外的值）：
import glob
import numpy as np
import os

output_file = "/home/lihaoxing/PA_IVF_532_25w_TAD/aggregateTADs/IVF_strength/IVF_strength_trimmed_mean.txt"

with open(output_file, "w") as out:
    out.write("Cell\tTrimmedMeanValue\n")
    for filename in glob.glob("/home/lihaoxing/PA_IVF_532_25w_TAD/aggregateTADs/IVF_strength/*.strength.txt"):
        values = []
        with open(filename) as f:
            for line in f:
                parts = line.strip().split("\t")
                try:
                    val = float(parts[4])
                    if np.isfinite(val) and val >= 0:  # 过滤掉 inf, -inf, 以及 < 0 的值
                        values.append(val)
                except ValueError:
                    continue
        if values:
            values = np.array(values)
            lower_bound, upper_bound = np.percentile(values, [2.5, 97.5])  # 计算 1% 和 99% 分位数
            filtered_values = values[(values >= lower_bound) & (values <= upper_bound)]  # 仅保留范围内的值
            trimmed_mean_value = np.mean(filtered_values) if len(filtered_values) > 0 else "NA"
        else:
            trimmed_mean_value = "NA"
        
        cell_name = os.path.basename(filename)  # 只保留文件名
        out.write(f"{cell_name}\t{trimmed_mean_value}\n")

print(f"截尾均值计算完成，结果保存在 {output_file}")
```

```python
#IVF
import pandas as pd
from scipy.stats import pearsonr
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from IPython.display import SVG, display

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# 读取数据文件
file_path = '/home/lihaoxing/TADstrength_vs_Time/IVF_strength_trimmed_mean_sort_time.txt'
data = pd.read_csv(file_path, sep='\t')

# 提取 rna 和 hic 列数据
rna = data['strength'].values
hic = data['time'].values


# 计算 Pearson 相关系数和 p 值
r_value, p_value = pearsonr(rna, hic)
print(r_value, p_value)
# 绘制散点图
plt.figure(figsize=(6, 6))
plt.scatter(rna, hic,color='black')
plt.title(f'IVF: TAD strength vs. Pseudotime\n R = {r_value:.3f}, p = {p_value:.3e}')
plt.ylabel('Pseudotime')
plt.xlabel('TAD Strength')
plt.grid(True)
# 保存图像为 PDF
plt.savefig('IVF_TADstrength_vs_time_corr.svg', dpi=300, bbox_inches='tight', transparent=True)
```

```python
#IVF 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import pairwise_distances_argmin_min

# 读取数据
tad_strength = pd.read_csv(
    "/home/lihaoxing/PA_IVF_532_25w_TAD/aggregateTADs/IVF_strength/individual_files/IVF_merged_tad_strength_matrix_new.txt",
    sep="\t", index_col=0
)

# 读取时间对应关系文件
time_df = pd.read_csv('/home/lihaoxing/PA_IVF_532_25w_TAD/aggregateTADs/IVF_strength/individual_files/ivf_time_sort.txt', sep='\t', header=None, names=['sample', 'time'])
sample_to_time = dict(zip(time_df['sample'], time_df['time']))

# 找到时间在12到14之间的样本
samples_to_remove = [sample for sample, time in sample_to_time.items() if 12.2 <= time <= 13.8]

# 丢掉时间在12到14之间的细胞
tad_strength = tad_strength.drop(columns=samples_to_remove, errors='ignore')

# ✅ 过滤掉 >0 的值少于 15 个的行
tad_strength = tad_strength[tad_strength.gt(0).sum(axis=1) >= 15]

# 将数据转换为 numpy 矩阵用于 KMeans
data = tad_strength.values

# 手肘法确定最佳聚类数
inertia = []
K_range = range(2, 11)
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init='auto').fit(data)
    inertia.append(kmeans.inertia_)

# 绘制手肘图
plt.figure(figsize=(6, 4))
plt.plot(K_range, inertia, 'o-')
plt.xlabel('Cluster number (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal k')
plt.grid(True)
plt.tight_layout()
plt.show()

from sklearn.metrics import silhouette_score

silhouette_scores = []
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init='auto').fit(data)
    score = silhouette_score(data, kmeans.labels_)
    silhouette_scores.append(score)

plt.figure(figsize=(6, 4))
plt.plot(K_range, silhouette_scores, 'o-')
plt.xlabel('Cluster number (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Analysis for Optimal k')
plt.grid(True)
plt.tight_layout()
plt.show()

#解释： 轮廓系数越高，聚类效果越好。它是衡量聚类紧密度和分离度的综合指标。
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from scipy.signal import savgol_filter

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# =============================
# 全局绘图参数
# =============================
plt.rcParams.update({
    'font.size': 18,
    'axes.titlesize': 20,
    'axes.labelsize': 20,
    'xtick.labelsize': 20,
    'ytick.labelsize': 20,
    'legend.fontsize': 20
})

# =============================
# 读取数据
# =============================
tad_strength = pd.read_csv(
    "/home/lihaoxing/PA_IVF_532_25w_TAD/aggregateTADs/IVF_strength/individual_files/IVF_merged_tad_strength_matrix_new.txt",
    sep="\t", index_col=0
)

time_df = pd.read_csv(
    "/home/lihaoxing/PA_IVF_532_25w_TAD/aggregateTADs/IVF_strength/individual_files/ivf_time_sort.txt",
    sep="\t", header=None, names=["sample", "time"]
)

sample_to_time = dict(zip(time_df["sample"], time_df["time"]))

# =============================
# 预处理
# =============================
# 去掉异常时间点
samples_to_remove = [s for s, t in sample_to_time.items() if 12.2 <= t <= 13.8]
tad_strength = tad_strength.drop(columns=samples_to_remove, errors='ignore')

# 保留有效信号（>=15个非0）
tad_strength = tad_strength[tad_strength.gt(0).sum(axis=1) >= 15]

# =============================
# KMeans 聚类（2类）
# =============================
kmeans = KMeans(n_clusters=2, random_state=99, n_init='auto')
tad_strength['cluster'] = kmeans.fit_predict(tad_strength)

# =============================
# 按平均强度重新定义 cluster（保证生物学意义）
# =============================
cluster_mean_strength = (
    tad_strength
    .drop(columns='cluster')
    .groupby(tad_strength['cluster'])
    .mean()
    .mean(axis=1)
)

# 高 → 低 排序
sorted_clusters = cluster_mean_strength.sort_values(ascending=False).index.tolist()

# 重新编号：高强度 = 0
cluster_remap = {old: new for new, old in enumerate(sorted_clusters)}
tad_strength['cluster'] = tad_strength['cluster'].map(cluster_remap)

# =============================
# cluster 顺序
# =============================
cluster_order = [0, 1]

# =============================
# 自动比例（推荐）
# =============================
total_display = 4538

cluster_counts = tad_strength['cluster'].value_counts(normalize=True)
cluster_ratio_map = {
    cid: cluster_counts.get(cid, 0)
    for cid in cluster_order
}

cluster_sizes = [int(total_display * cluster_ratio_map[c]) for c in cluster_order]
cluster_sizes[-1] = total_display - sum(cluster_sizes[:-1])

# =============================
# cluster 名称
# =============================
cluster_name_map = {
    0: "HTP",
    1: "LTP"
}

# =============================
# cluster 分组 & 排序
# =============================
clustered_data = []
cluster_boundaries = []
cluster_labels_pos = []

start = 0

for i, cid in enumerate(cluster_order):
    block = tad_strength[tad_strength['cluster'] == cid].drop(columns='cluster')

    # 行排序（增强结构）
    block = block.assign(row_mean=block.mean(axis=1)) \
                 .sort_values('row_mean') \
                 .drop(columns='row_mean')

    n = min(cluster_sizes[i], len(block))
    block = block.iloc[:n]

    clustered_data.append(block)

    start += n
    cluster_boundaries.append(start)
    cluster_labels_pos.append(start - n / 2)

# 合并
data_sorted = pd.concat(clustered_data)

# =============================
# 绘图（仅主热图）
# =============================
fig, ax_heat = plt.subplots(figsize=(6, 5))

cmap = sns.diverging_palette(240, 10, as_cmap=True)

sns.heatmap(
    data_sorted,
    cmap=cmap,
    vmin=0,
    vmax=6,
    yticklabels=False,
    xticklabels=False,
    cbar_kws={'label': 'TAD Strength'},
    ax=ax_heat
)



# 分割线
for y in cluster_boundaries[:-1]:
    ax_heat.axhline(y=y, color='white', linewidth=3)

# 标签
for pos, cid in zip(cluster_labels_pos, cluster_order):
    ax_heat.text(
        x=-5,
        y=pos,
        s=cluster_name_map[cid],
        va='center',
        ha='right',
        fontsize=20
    )

ax_heat.set_ylabel("")
ax_heat.set_title("TAD Strength Dynamics (IVF)", fontsize=20)
ax_heat.set_xlabel("Pseudotime", fontsize=20)   # 添加 x 轴标签

# 颜色条字体大小
cbar = ax_heat.collections[0].colorbar
cbar.ax.tick_params(labelsize=14)

# =============================
# 保存
# =============================
plt.tight_layout()
plt.savefig(
    "IVF_TAD_heatmap_2clusters.png",
    dpi=300,
    bbox_inches='tight',
    transparent=True
)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# ------------------------- 读取时间信息 -------------------------
time_df = pd.read_csv(
    '/home/lihaoxing/PA_IVF_532_25w_TAD/aggregateTADs/IVF_strength/individual_files/ivf_time_sort.txt',
    sep='\t', header=None, names=['sample', 'time']
)
sample_to_time = dict(zip(time_df['sample'], time_df['time']))

# ------------------------- 参数 -------------------------
optimal_k = 2
bin_width = 0.1
bins = np.arange(0, 14 + bin_width, bin_width)   # 0, 0.1, 0.2, ..., 14.0
bin_centers = (bins[:-1] + bins[1:]) / 2         # 每个箱的中心点

fig, ax = plt.subplots(figsize=(6, 5))
colors_global = [plt.cm.tab10(i) for i in range(optimal_k)]

# ------------------------- 绘制 Global Trend（分箱均值±标准差） -------------------------
for i in range(optimal_k):
    cluster_data = tad_strength[tad_strength['cluster'] == i].drop(columns='cluster')
    mean_values = cluster_data.mean(axis=0)   # 每个样本的平均 TAD 强度

    # 收集时间点和对应的强度值
    time_vals = []
    values = []
    for s in mean_values.index:
        if s in sample_to_time:
            time_vals.append(sample_to_time[s])
            values.append(mean_values[s])
    
    time_vals = np.array(time_vals)
    values = np.array(values)
    
    # 计算原始数据的相关性（仍然基于所有点）
    if len(time_vals) >= 2:
        r, _ = pearsonr(time_vals, values)
        print(f"Cluster {i} 相关性 r = {r:.4f}")
    else:
        print(f"警告：Cluster {i} 有效数据点不足2个（{len(time_vals)}个），无法计算相关性")
    
    # 分箱：对每个 bin 计算均值和标准差
    means = []
    stds = []
    for j in range(len(bins) - 1):
        left, right = bins[j], bins[j+1]
        # 落在该箱内的点（包括左闭右开，但最后一个箱包含右端点）
        if j == len(bins) - 2:
            mask = (time_vals >= left) & (time_vals <= right)
        else:
            mask = (time_vals >= left) & (time_vals < right)
        bin_vals = values[mask]
        if len(bin_vals) > 0:
            means.append(np.mean(bin_vals))
            stds.append(np.std(bin_vals))
        else:
            means.append(np.nan)
            stds.append(np.nan)
    
    means = np.array(means)
    stds = np.array(stds)
    
    # 去掉全是 NaN 的 bin（不画线）
    valid = ~np.isnan(means)
    bin_centers_valid = bin_centers[valid]
    means_valid = means[valid]
    stds_valid = stds[valid]
    
    label = "RRGP" if i == 0 else "SRGP"
    color = colors_global[i]
    
    # 绘制均值曲线
    if len(bin_centers_valid) > 0:
        ax.plot(bin_centers_valid, means_valid, color=color, linewidth=2, label=label)
        # 绘制标准差阴影（均值±标准差）
        ax.fill_between(bin_centers_valid,
                         means_valid - stds_valid,
                         means_valid + stds_valid,
                         color=color, alpha=0.2)
    else:
        print(f"警告：Cluster {i} 分箱后无有效数据，跳过绘图")

ax.set_title('IVF Cells — TAD Strength Trend', fontsize=20)
ax.set_xlabel('Pseudotime', fontsize=20)
ax.set_ylabel('Average TAD Strength', fontsize=20)
ax.set_xlim(0, 14)
ax.legend(fontsize=18)

plt.savefig('IVF_TADstrength_global_curve.svg', dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```

```python
import os
import glob
from collections import defaultdict

# ================== 配置 ==================
GENE_BED_FILES = {
    'ZGA': '/home/lihaoxing/scRNA_expression_vs_pseudotime/IVF_ZGA.bed',
    'MD': '/home/lihaoxing/scRNA_expression_vs_pseudotime/IVF_MD.bed',
    'other': '/home/lihaoxing/scRNA_expression_vs_pseudotime/IVF_other.bed'
}

OUTPUT_DIR = 'split_bedpe'
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ================== 读取基因区间 ==================
def read_bed_genes(bed_path):
    """返回列表，每个元素为 (chr, start, end)（start, end 为整数）"""
    intervals = []
    with open(bed_path) as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) < 3:
                continue
            chrom = parts[0]
            try:
                start = int(parts[1])
                end = int(parts[2])
            except ValueError:
                continue
            intervals.append((chrom, start, end))
    return intervals

gene_intervals = {name: read_bed_genes(path) for name, path in GENE_BED_FILES.items()}

# ================== 重叠检测 ==================
def anchor_overlaps_gene(chrom, start, end, gene_list):
    """检查单个锚点 (chrom, start, end) 是否与 gene_list 中任一区间有重叠"""
    for g_chrom, g_start, g_end in gene_list:
        if chrom == g_chrom and start < g_end and end > g_start:
            return True
    return False

def bedpe_overlaps_gene(line, gene_list):
    """检查 bedpe 行是否与基因区间重叠（任一锚点）"""
    parts = line.strip().split()
    if len(parts) < 6:
        return False
    # 锚点1
    if anchor_overlaps_gene(parts[0], int(parts[1]), int(parts[2]), gene_list):
        return True
    # 锚点2
    if anchor_overlaps_gene(parts[3], int(parts[4]), int(parts[5]), gene_list):
        return True
    return False

# ================== 处理所有 .bedpe 文件 ==================
for bedpe_path in glob.glob('*.bedpe'):
    base = os.path.splitext(bedpe_path)[0]
    print(f'Processing {base}...')
    
    with open(bedpe_path) as f:
        lines = f.readlines()
    
    # 按优先级分配：ZGA > MD > other
    zga_lines = []
    md_lines = []
    other_lines = []
    
    # 先筛选 ZGA
    for line in lines:
        if bedpe_overlaps_gene(line, gene_intervals['ZGA']):
            zga_lines.append(line)
    
    # 剩余中筛选 MD
    remaining = [line for line in lines if line not in zga_lines]
    for line in remaining:
        if bedpe_overlaps_gene(line, gene_intervals['MD']):
            md_lines.append(line)
    
    # 剩下的归 other（不包含 ZGA 和 MD 重叠的）
    remaining2 = [line for line in remaining if line not in md_lines]
    other_lines = remaining2
    
    # 写入输出文件
    with open(os.path.join(OUTPUT_DIR, f'{base}_ZGA.bedpe'), 'w') as f:
        f.writelines(zga_lines)
    with open(os.path.join(OUTPUT_DIR, f'{base}_MD.bedpe'), 'w') as f:
        f.writelines(md_lines)
    with open(os.path.join(OUTPUT_DIR, f'{base}_other.bedpe'), 'w') as f:
        f.writelines(other_lines)

print("All done. Results in", OUTPUT_DIR)
```

```python
import os
import glob
from collections import defaultdict

# ================== 配置 ==================
GENE_BED = '/home/lihaoxing/scRNA_expression_vs_pseudotime/IVF_ZGA.bed'   # ZGA 基因区间
INPUT_DIR = 'split_bedpe'          # 存放 *_ZGA.bedpe 的目录（每个细胞一个）
OUTPUT_DIR = 'per_cell_gene_bedpe' # 输出目录
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ================== 读取基因区间 ==================
gene_intervals = {}  # gene_id -> (chrom, start, end)
with open(GENE_BED) as f:
    for line in f:
        parts = line.strip().split()
        if len(parts) < 4:
            continue
        chrom, start, end, gene_id = parts[0], int(parts[1]), int(parts[2]), parts[3]
        gene_intervals[gene_id] = (chrom, start, end)
print(f"Loaded {len(gene_intervals)} genes.")

# ================== 重叠检测函数 ==================
def anchor_overlaps(anchor_chrom, anchor_start, anchor_end, gene_chrom, gene_start, gene_end):
    return anchor_chrom == gene_chrom and anchor_start < gene_end and anchor_end > gene_start

def get_overlapping_genes(parts, gene_intervals):
    """返回与该 loop 行重叠的所有基因 ID 列表"""
    chrom1, start1, end1 = parts[0], int(parts[1]), int(parts[2])
    chrom2, start2, end2 = parts[3], int(parts[4]), int(parts[5])
    overlapping = []
    for gene_id, (g_chrom, g_start, g_end) in gene_intervals.items():
        if anchor_overlaps(chrom1, start1, end1, g_chrom, g_start, g_end) or \
           anchor_overlaps(chrom2, start2, end2, g_chrom, g_start, g_end):
            overlapping.append(gene_id)
    return overlapping

# ================== 处理每个细胞的 ZGA.bedpe ==================
# 获取所有以 _ZGA.bedpe 结尾的文件
zga_files = glob.glob(os.path.join(INPUT_DIR, '*_ZGA.bedpe'))
print(f"Found {len(zga_files)} ZGA files.")

for zga_path in zga_files:
    # 提取细胞名（例如 IVF-2-168-1）
    basename = os.path.basename(zga_path)
    cell_id = basename.replace('_ZGA.bedpe', '')   # 去掉 _ZGA.bedpe 后缀
    
    print(f"Processing {cell_id} ...")
    
    # 用字典暂存每个基因的 loop 行列表
    gene_loops = defaultdict(list)
    
    with open(zga_path) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            parts = line.split()
            if len(parts) < 6:
                continue
            genes = get_overlapping_genes(parts, gene_intervals)
            if not genes:
                continue   # 跳过不与任何基因重叠的 loop（或可选择写入 “unassigned”）
            for gene_id in genes:
                gene_loops[gene_id].append(line)
    
    # 写入该细胞每个基因对应的文件（仅当有 loop 时才创建）
    for gene_id, loops in gene_loops.items():
        if not loops:
            continue
        out_filename = f"{cell_id}_{gene_id}.bedpe"
        out_path = os.path.join(OUTPUT_DIR, out_filename)
        with open(out_path, 'w') as out_f:
            out_f.write('\n'.join(loops) + ('\n' if loops else ''))

print("All done. Results in", OUTPUT_DIR)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from scipy.signal import savgol_filter

# =============================
# 全局绘图参数
# =============================
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'

plt.rcParams.update({
    'font.size': 18,
    'axes.titlesize': 20,
    'axes.labelsize': 20,
    'xtick.labelsize': 20,
    'ytick.labelsize': 20,
    'legend.fontsize': 20
})

# =============================
# 读取数据
# =============================
tad_strength = pd.read_csv(
    "/home/lihaoxing/PA_IVF_532_25w_TAD/aggregateTADs/PA_strength/individual_files/PA_merged_tad_strength_matrix_new.txt",
    sep="\t", index_col=0
)

time_df = pd.read_csv(
    "/home/lihaoxing/PA_IVF_532_25w_TAD/aggregateTADs/PA_strength/individual_files/pa_time_sort.txt",
    sep="\t", header=None, names=["sample", "time"]
)

sample_to_time = dict(zip(time_df["sample"], time_df["time"]))

# =============================
# 预处理
# =============================
# 去掉异常时间点
samples_to_remove = [s for s, t in sample_to_time.items() if 12.2 <= t <= 13.8]
tad_strength = tad_strength.drop(columns=samples_to_remove, errors='ignore')

# 保留有效信号（>=15个非0）
tad_strength = tad_strength[tad_strength.gt(0).sum(axis=1) >= 15]

# =============================
# KMeans 聚类（2类）
# =============================
kmeans = KMeans(n_clusters=2, random_state=99, n_init='auto')
tad_strength['cluster'] = kmeans.fit_predict(tad_strength)

# =============================
# 按平均强度重新定义 cluster（保证生物学意义）
# =============================
cluster_mean_strength = (
    tad_strength
    .drop(columns='cluster')
    .groupby(tad_strength['cluster'])
    .mean()
    .mean(axis=1)
)

# 高 → 低 排序
sorted_clusters = cluster_mean_strength.sort_values(ascending=False).index.tolist()

# 重新编号：高强度 = 0
cluster_remap = {old: new for new, old in enumerate(sorted_clusters)}
tad_strength['cluster'] = tad_strength['cluster'].map(cluster_remap)

# =============================
# cluster 顺序
# =============================
cluster_order = [0, 1]

# =============================
# 自动比例（推荐）
# =============================
total_display = 4538

cluster_counts = tad_strength['cluster'].value_counts(normalize=True)
cluster_ratio_map = {
    cid: cluster_counts.get(cid, 0)
    for cid in cluster_order
}

cluster_sizes = [int(total_display * cluster_ratio_map[c]) for c in cluster_order]
cluster_sizes[-1] = total_display - sum(cluster_sizes[:-1])

# =============================
# cluster 名称
# =============================
cluster_name_map = {
    0: "HTP",
    1: "LTP"
}

# =============================
# cluster 分组 & 排序
# =============================
clustered_data = []
cluster_means = []
cluster_boundaries = []
cluster_labels_pos = []

start = 0

for i, cid in enumerate(cluster_order):
    block = tad_strength[tad_strength['cluster'] == cid].drop(columns='cluster')

    # 行排序（增强结构）
    block = block.assign(row_mean=block.mean(axis=1)) \
                 .sort_values('row_mean') \
                 .drop(columns='row_mean')

    n = min(cluster_sizes[i], len(block))
    block = block.iloc[:n]

    clustered_data.append(block)

    # ===== 平滑均值曲线 =====
    mean_profile = block.mean(axis=0)

    window = min(200, len(mean_profile))
    if window % 2 == 0:
        window -= 1
    window = max(window, 3)

    smoothed = savgol_filter(mean_profile.values, window_length=window, polyorder=2)
    cluster_means.append(smoothed)

    start += n
    cluster_boundaries.append(start)
    cluster_labels_pos.append(start - n / 2)

# 合并
data_sorted = pd.concat(clustered_data)

# =============================
# 绘图（仅保留主热图）
# =============================
fig = plt.figure(figsize=(8, 6))          # 调整为单图合适的高度
ax_heat = fig.add_subplot(111)

cmap = sns.diverging_palette(240, 10, as_cmap=True)

sns.heatmap(
    data_sorted,
    cmap=cmap,
    vmin=0,
    vmax=6,
    yticklabels=False,
    xticklabels=False,
    cbar_kws={'label': 'TLD Strength'},
    ax=ax_heat
)

# 分割线
for y in cluster_boundaries[:-1]:
    ax_heat.axhline(y=y, color='white', linewidth=3)

# 标签
for pos, cid in zip(cluster_labels_pos, cluster_order):
    ax_heat.text(
        x=-5,
        y=pos,
        s=cluster_name_map[cid],
        va='center',
        ha='right',
        fontsize=20
    )

ax_heat.set_ylabel("")
ax_heat.set_xlabel("Pseudotime", fontsize=20)
ax_heat.set_title("TLD Strength Dynamics (PA)", fontsize=20)

# 颜色条字体大小
cbar = ax_heat.collections[0].colorbar
cbar.ax.tick_params(labelsize=14)

# =============================
# 保存
# =============================
plt.tight_layout()
plt.savefig(
    "/home/lihaoxing/jupyter/PA_TLD_heatmap_2clusters.png",
    dpi=300,
    bbox_inches='tight',
    transparent=True
)

plt.show()
```

```python
#gene 2  windows 
import pandas as pd
import glob

# =============================
# Step 1：从 window.txt 构建初始 cluster
# =============================
def build_cluster_bed(prefix):
    files = glob.glob(f"{prefix}_cluster_*_window.txt")

    dfs = []
    for f in files:
        cluster_id = int(f.split("_cluster_")[1].split("_")[0])
        df = pd.read_csv(f, sep="\t", header=None)
        df['cluster'] = cluster_id
        dfs.append(df)

    data = pd.concat(dfs, ignore_index=True)

    data.columns = [
        'chr', 'start', 'end', 'window_id',
        'chr_gene', 'gene_start', 'gene_end', 'gene_id',
        'cluster'
    ]

    # =============================
    # window → cluster 频率统计
    # =============================
    freq = (
        data.groupby(['window_id', 'cluster'])
        .size()
        .reset_index(name='count')
    )

    # count 最大优先，tie 时 cluster 小优先
    freq = freq.sort_values(['window_id', 'count', 'cluster'], ascending=[True, False, True])
    window_cluster_map = freq.drop_duplicates('window_id')

    # =============================
    # 合并回原数据
    # =============================
    data = data.merge(
        window_cluster_map[['window_id', 'cluster']],
        on='window_id',
        suffixes=('', '_final')
    )

    data_final = data[data['cluster'] == data['cluster_final']]

    # 每个 window 保留一条记录
    bed_df = (
        data_final
        .drop_duplicates(subset=['window_id'])
        [['chr', 'start', 'end', 'window_id', 'cluster_final']]
        .rename(columns={'cluster_final': 'cluster'})
    )

    return bed_df


# =============================
# Step 2：补全 missing windows → cluster2
# =============================
def fill_missing_windows(prefix, bed_df):
    # 全集
    all_bins = pd.read_csv("500k_bin.bed", sep="\t", header=None)
    all_bins.columns = ['chr', 'start', 'end', 'window_id']

    # merge（用 window_id，最安全）
    merged = pd.merge(
        all_bins,
        bed_df[['window_id', 'cluster']],
        on='window_id',
        how='left'
    )

    # missing
    missing = merged[merged['cluster'].isna()].copy()

    # 补到 cluster2
    missing['cluster'] = 2

    # 合并
    final = pd.concat([
        merged[~merged['cluster'].isna()],
        missing
    ], ignore_index=True)

    # =============================
    # 输出（带 window_id）
    # =============================
    for cid in [0, 1, 2]:
        out = final[final['cluster'] == cid][['chr', 'start', 'end', 'window_id']]
        out_file = f"{prefix}_cluster_{cid}_.bed"
        out.to_csv(out_file, sep="\t", header=False, index=False)

        print(f"✔ Saved: {out_file} ({len(out)} regions)")

    print(f"{prefix} total windows:", len(final))
    print("-" * 40)


# =============================
# 主流程
# =============================
def process(prefix):
    bed_df = build_cluster_bed(prefix)
    fill_missing_windows(prefix, bed_df)


# =============================
# 运行
# =============================
process("IVF")
process("PA")

def report_proportion(prefix):
    total = 4538
    print(f"\n=== {prefix} proportion ===")

    for cid in [0, 1, 2]:
        f = f"{prefix}_cluster_{cid}.bed"
        df = pd.read_csv(f, sep="\t", header=None)
        n = len(df)
        ratio = n / total

        print(f"Cluster {cid}: {n} ({ratio:.4f})")

# 运行
report_proportion("IVF")
report_proportion("PA")
```

```python
def chromosome_distribution(prefix):
    print(f"\n=== {prefix} chromosome distribution ===")

    all_df = []

    for cid in [0, 1]:
        f = f"{prefix}_cluster_{cid}_random99.bed"
        df = pd.read_csv(f, sep="\t", header=None)
        df.columns = ['chr', 'start', 'end', 'window_id']
        df['cluster'] = cid
        all_df.append(df)

    df_all = pd.concat(all_df)

    # =============================
    # 每个 chr 内 cluster 比例
    # =============================
    chr_counts = (
        df_all.groupby(['chr', 'cluster'])
        .size()
        .reset_index(name='count')
    )

    # 转成比例
    chr_total = chr_counts.groupby('chr')['count'].transform('sum')
    chr_counts['ratio'] = chr_counts['count'] / chr_total

    print(chr_counts)

    # 保存
    chr_counts.to_csv(f"{prefix}_chr_distribution.txt", sep="\t", index=False)
```

```python
import pandas as pd
import glob
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator

plt.rcParams['font.family'] = 'DejaVu Sans'

# =============================
# Step 1：正确计算 gene density（只保留 cluster 0 和 1）
# =============================
def compute_gene_density(prefix):

    # ===== gene mapping =====
    files = glob.glob(f"{prefix}_cluster_*_window_2cluster_random99.txt")
    dfs = []
    for f in files:
        df = pd.read_csv(f, sep="\t", header=None)
        df.columns = [
            'chr','start','end','window_id',
            'chr_gene','gene_start','gene_end','gene_id'
        ]
        dfs.append(df)
    gene_map = pd.concat(dfs, ignore_index=True)

    # 去重
    gene_map = gene_map.drop_duplicates(['window_id', 'gene_id'])

    # ===== 正确 cluster（来自 BED）=====
    cluster_list = []
    for cid in [0,1]:
        df = pd.read_csv(f"{prefix}_cluster_{cid}.bed", sep="\t", header=None)
        df.columns = ['chr','start','end','window_id']
        df['cluster'] = cid
        cluster_list.append(df[['window_id','cluster']])
    cluster_df = pd.concat(cluster_list, ignore_index=True)

    # 检查是否有重复 window_id（同一个 window 出现在两个 cluster）
    dup_check = cluster_df.groupby('window_id').size().max()
    if dup_check > 1:
        raise ValueError(f"window_id 出现在多个 cluster 中，请检查！最大重复次数: {dup_check}")

    # ===== merge =====
    merged = pd.merge(gene_map, cluster_df, on='window_id', how='inner')
    merged['cluster'] = merged['cluster'].astype(int)

    # ===== gene density =====
    df = (
        merged[merged['cluster'].isin([0,1])]   # 只保留 0 和 1
        .groupby(['window_id','cluster'])
        .size()
        .reset_index(name='gene_count')
    )

    return df

# =============================
# Step 2：绘图
# =============================
def plot_gene_density(prefix):
    df = compute_gene_density(prefix)

    plt.figure(figsize=(5,4))
    order = [0,1]
    palette = ['#1f77b4', '#ff7f0e']

    ax = sns.violinplot(
        data=df,
        x='cluster',
        y='gene_count',
        order=order,
        palette=palette,
        cut=0
    )

    # =============================
    # 统计检验
    # =============================
    pairs = [(0,1)]
    annotator = Annotator(
        ax,
        pairs,
        data=df,
        x='cluster',
        y='gene_count',
        order=order
    )
    annotator.configure(
        test='Mann-Whitney',
        text_format='star',
        loc='outside',
        verbose=1
    )
    annotator.apply_and_annotate()

    # =============================
    # 美化
    # =============================
    plt.xlabel("Cluster", fontsize=12)
    plt.ylabel("Gene count per window", fontsize=12)
    plt.title(f"{prefix} gene density", fontsize=14, pad=70)

    plt.tight_layout()
    plt.savefig(f"{prefix}_gene_density.pdf", bbox_inches='tight')
    plt.show()

# =============================
# Step 3：summary
# =============================
def gene_density_summary(prefix):
    df = compute_gene_density(prefix)
    summary = df.groupby('cluster')['gene_count'].agg(['mean','median','std'])
    print(f"\n=== {prefix} gene density summary ===")
    print(summary)

# =============================
# 运行
# =============================
plot_gene_density("IVF")
plot_gene_density("PA")

gene_density_summary("IVF")
gene_density_summary("PA")
```

```python
import pandas as pd
import glob
import os

# =============================
# Step：根据 ID 提取 bed
# =============================
def extract_debin_by_id(bed_file, id_file, output_file):

    # 读取 bed
    bed_df = pd.read_csv(bed_file, sep="\t", header=None)
    bed_df.columns = ['chr', 'start', 'end', 'window_id']

    # 读取 ID
    ids = pd.read_csv(id_file, header=None)[0].tolist()
    ids = set(ids)  # 提速

    # 筛选
    filtered = bed_df[bed_df['window_id'].isin(ids)]

    # 输出
    filtered.to_csv(output_file, sep="\t", header=False, index=False)

    print(f"✔ {os.path.basename(output_file)}: {len(filtered)} regions")


# =============================
# 自动匹配 IVF 2C / 4C
# =============================
def process_all():

    bed_files = glob.glob("dedoc2_*_debin.bed")

    for bed_file in bed_files:

        prefix = bed_file.replace("_debin.bed", "")

        for tag in ["H1", "H2"]:

            id_file = f"{prefix}_{tag}_debin_id.txt"

            if not os.path.exists(id_file):
                print(f"⚠ Missing: {id_file}")
                continue

            out_file = f"{prefix}_{tag}.bed"

            extract_debin_by_id(bed_file, id_file, out_file)


# =============================
# 运行
# =============================
process_all()
```
