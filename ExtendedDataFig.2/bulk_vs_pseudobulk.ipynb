```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
import numpy as np

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# 读取文件（确认表头存在）
df = pd.read_csv('PA_IVF_4C_PC1.bedGraph', sep='\t')

# 强制转换为数值，无法转换的变为 NaN
for col in ['IVF_bulk', 'IVF_pseudobulk']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 移除含有 NaN 或 inf 的行
df_clean = df.replace([np.inf, -np.inf], np.nan).dropna(subset=['IVF_bulk', 'IVF_pseudobulk'])

# 告知清除了多少异常行
print(f"原始行数: {len(df)}, 清理后行数: {len(df_clean)}")

x = df_clean['IVF_bulk']
y = df_clean['IVF_pseudobulk']

r, p = pearsonr(x, y)
print(f"Pearson r = {r:.4f}, p = {p:.2e}")

plt.figure(figsize=(6, 6))
sns.scatterplot(x=x, y=y, s=10, color='steelblue', alpha=0.6)
plt.xlabel('IVF 4C bulk', fontsize=14)
plt.ylabel('IVF 4C pseudobulk', fontsize=14)
plt.title(f'PC1 Pearson Correlation: r = {r:.2f}', fontsize=14)
plt.savefig('IVF_4C_AB_bulk_vs_pseudobulk_scatter_new.svg', dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```
