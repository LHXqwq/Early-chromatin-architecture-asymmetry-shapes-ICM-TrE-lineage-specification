```python
import matplotlib.pyplot as plt

# =========================
# 全局参数
# =========================
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

# =========================
# 数据
# =========================
labels = ['PA-2C', 'PA-4C', 'IVF-2C', 'IVF-4C']

iDEG_data = [618, 1181, 354, 2116]
iDET_data = [27616, 20295, 33804, 18242]

colors = ['#3cb44b', '#e6194b', '#3cb44b', '#e6194b']

# =========================
# 创建子图（1行2列）
# =========================
fig, axes = plt.subplots(1, 2, figsize=(10, 6))

# =========================
# 图1：iDEGs
# =========================
ax1 = axes[0]
ax1.bar(labels, iDEG_data, color=colors)
ax1.set_title("Number of iDEGs", fontsize=20)
ax1.set_ylabel("Counts", fontsize=20)
ax1.set_ylim(0, 2300)
ax1.tick_params(axis='x', labelsize=15)
ax1.tick_params(axis='y', labelsize=15)

# 标注数值
for i, value in enumerate(iDEG_data):
    ax1.text(i, value + 50, str(value), ha='center', fontsize=12)

# =========================
# 图2：iDETs
# =========================
ax2 = axes[1]
ax2.bar(labels, iDET_data, color=colors)
ax2.set_title("Number of iDETs", fontsize=20)
ax2.set_ylim(0, 36000)
ax2.tick_params(axis='x', labelsize=15)
ax2.tick_params(axis='y', labelsize=15)

# 标注数值
for i, value in enumerate(iDET_data):
    ax2.text(i, value + 500, str(value), ha='center', fontsize=12)

# =========================
# 布局 & 保存
# =========================
plt.tight_layout()
plt.savefig('iDEG_iDET_combined.svg', dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```
