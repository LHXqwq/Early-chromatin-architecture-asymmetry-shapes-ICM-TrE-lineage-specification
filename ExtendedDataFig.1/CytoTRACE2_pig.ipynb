```python
#pre-implant_mouse_embryo
#scRNAseq
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator
from scipy import stats

# 读取数据
data = pd.read_csv('pre-implant_mouse_embryo_RPKMCytoTRACE_plot_table.txt', sep='\t')

# 设置绘图风格
sns.set(style="whitegrid")
plt.rcParams['font.sans-serif'] = ['DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

trajectory_1 = [
    "early2cell", "mid2cell","late2cell", "4cell","8cell","16cell",   
    "earlyblast","midblast","lateblast"
]

# 合并两个轨迹顺序，轨迹1在前，轨迹2在后
ordered_phenotypes = trajectory_1

# 调整图形的宽度
fig, ax = plt.subplots(figsize=(15, 10))  # 增加宽度

# 创建小提琴图并应用颜色映射和排序
sns.violinplot(x='Phenotype', y='CytoTRACE', data=data,  order=ordered_phenotypes,
               ax=ax, inner="quart", cut=0, scale="width")
sns.stripplot(x='Phenotype', y='CytoTRACE', data=data, order=ordered_phenotypes, 
              color='black', size=5, jitter=True)  # 添加jitter

# 配置统计检验
pairs = [("early2cell", "mid2cell"),("mid2cell", "late2cell"),("late2cell", "4cell"),("4cell", "8cell"),("8cell", "16cell"),
         ("16cell", "earlyblast"),("earlyblast", "midblast"),("midblast", "lateblast")]
annotator = Annotator(ax, pairs, data=data, x="Phenotype", y="CytoTRACE", order=ordered_phenotypes)

# 选择统计检验方法
chosen_test = 'Mann-Whitney'  # 你可以动态设置
annotator.configure(test=chosen_test, text_format='star', line_height=0.01, 
                    line_width=1, loc='outside', fontsize=20)
annotator.apply_and_annotate()

# 旋转 x 轴标签
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.ylabel('CytoTRACE', fontsize=20)
plt.xlabel('')
plt.tight_layout()  # 调整布局以避免标签被剪裁

# 保存图片为 PDF
# plt.savefig('project_contacts.pdf', format='pdf')

# 显示图形
plt.show()
```

```python
#scRNAseq，原始reads需要按照这个比例降采样
#Uniquely mapped read pairs number
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator
from scipy import stats

# 读取数据
data = pd.read_csv('count_bam.txt', sep='\t')

# 将PA-MII的Uniquely_mapped_reads_number减少10%
data.loc[data['celltype'] == 'MII', 'Uniquely_mapped_reads_number'] *= 0.35
data.loc[data['celltype'] == 'PA-MII-new', 'Uniquely_mapped_reads_number'] *= 0.5
data.loc[data['celltype'] == 'PA-MII', 'Uniquely_mapped_reads_number'] *= 0.6
data.loc[data['celltype'] == 'PA-2C', 'Uniquely_mapped_reads_number'] *= 0.9
data.loc[data['celltype'] == 'IVF-4C', 'Uniquely_mapped_reads_number'] *= 0.6

# 设置绘图风格
sns.set(style="whitegrid")
plt.rcParams['font.sans-serif'] = ['DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 定义配色
color_mapping = {
    "MII": "#ffcc00",          # 新增: MII
    "PA-MII": "#ffe119",
    "PA-MII-new": "#ffd700",   # 新增: PA-MII-new
    "PA-2C": "#3cb44b",
    "PA-4C": "#e6194b",
    "IVF-Zygote": "#ffe119",  
    "IVF-2C": "#3cb44b",
    "IVF-4C": "#e6194b",
    "Other": "#a9a9a9"
}


# 轨迹信息
trajectory_1 = ["MII","PA-MII-new","PA-MII", "PA-2C", "PA-4C"]
trajectory_2 = ["IVF-Zygote","IVF-2C", "IVF-4C"]
ordered_phenotypes = trajectory_1 + trajectory_2

# 调整图形的宽度
fig, ax = plt.subplots(figsize=(15, 6))

# 创建小提琴图并应用颜色映射和排序
sns.violinplot(x='celltype', y='Uniquely_mapped_reads_number', data=data, 
               palette=color_mapping, order=ordered_phenotypes,
               ax=ax, inner="quart", cut=0, scale="width")
sns.stripplot(x='celltype', y='Uniquely_mapped_reads_number', data=data, order=ordered_phenotypes, 
              color='black', size=5, jitter=True)  # 添加jitter

# 配置统计检验
pairs = [("MII", "PA-MII-new"),("PA-MII-new", "PA-MII"),("PA-MII", "PA-2C"),("PA-2C", "PA-4C"),("IVF-Zygote", "IVF-2C"),("IVF-2C", "IVF-4C")]
annotator = Annotator(ax, pairs, data=data, x="celltype", y="Uniquely_mapped_reads_number", order=ordered_phenotypes)

# 选择统计检验方法
chosen_test = 'Mann-Whitney'  # 你可以动态设置
annotator.configure(test=chosen_test, text_format='star', line_height=0.01, 
                    line_width=1, loc='outside', fontsize=20)
annotator.apply_and_annotate()

# 旋转 x 轴标签
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.ylabel('All read number', fontsize=20)
plt.xlabel('Cell Type', fontsize=20)  # 增加x轴标签
plt.tight_layout()

# 保存图片为 PDF
# plt.savefig('project_contacts.pdf', format='pdf')

# 显示图形
plt.show()
```

```python
# dip - IVF only
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator
from IPython.display import SVG, display

# 字体设置
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 读取数据
data = pd.read_csv('PA_IVF_198_CytoTRACE2_results_cluster2_dip.txt', sep='\t')

# ID 替换：旧 ID -> 新 ID
id_mapping = {
    "IVF-2C-iDET1": "IVF-iDET-2α",
    "IVF-2C-iDET2": "IVF-iDET-2β",
    "IVF-4C-iDET1": "IVF-iDET-4α",
    "IVF-4C-iDET2": "IVF-iDET-4β",
}

# 如果数据中 x 列仍为旧 ID，则替换为新 ID
data['DNA-hAT-Charlie'] = data['DNA-hAT-Charlie'].replace(id_mapping)

# 原始配色映射
color_mapping_raw = {
    "PA-2C-iDET1": "#7dace4",
    "PA-2C-iDET2": "#ff165d",
    "PA-4C-iDET1": "green",
    "PA-4C-iDET2": "orange",
    "IVF-2C-iDET1": "#fb9795",
    "IVF-2C-iDET2": "#E9B44C",
    "IVF-4C-iDET1": "#7c5e8c",
    "IVF-4C-iDET2": "#b0e7ea",
}

# 将配色映射到新的 IVF ID，颜色沿用 color_mapping_raw
color_mapping = {
    "PA-2C-iDET1": color_mapping_raw["PA-2C-iDET1"],
    "PA-2C-iDET2": color_mapping_raw["PA-2C-iDET2"],
    "PA-4C-iDET1": color_mapping_raw["PA-4C-iDET1"],
    "PA-4C-iDET2": color_mapping_raw["PA-4C-iDET2"],
    "IVF-iDET-2α": color_mapping_raw["IVF-2C-iDET1"],
    "IVF-iDET-2β": color_mapping_raw["IVF-2C-iDET2"],
    "IVF-iDET-4α": color_mapping_raw["IVF-4C-iDET1"],
    "IVF-iDET-4β": color_mapping_raw["IVF-4C-iDET2"],
}

# 横坐标标签替换
x_label_mapping = {
    "IVF-iDET-2α": "IVF-iDET-2α",
    "IVF-iDET-2β": "IVF-iDET-2β",
    "IVF-iDET-4α": "IVF-iDET-4α",
    "IVF-iDET-4β": "IVF-iDET-4β",
}

# IVF 轨迹，使用新 ID
trajectory_2 = [
    "IVF-iDET-2α",
    "IVF-iDET-2β",
    "IVF-iDET-4α",
    "IVF-iDET-4β",
]
ordered_phenotypes = trajectory_2

# 绘图风格
sns.set(style="whitegrid")

# 创建画布
fig, ax = plt.subplots(figsize=(10, 6))

# 绘制小提琴图和散点
sns.violinplot(
    x='DNA-hAT-Charlie',
    y='CytoTRACE2_Score',
    data=data,
    palette=color_mapping,
    order=ordered_phenotypes,
    ax=ax,
    inner="quart",
    cut=0,
    scale="width"
)

sns.stripplot(
    x='DNA-hAT-Charlie',
    y='CytoTRACE2_Score',
    data=data,
    order=ordered_phenotypes,
    color='black',
    size=5
)

# 每对比较及对应检验方式，使用新 ID
pairs_with_tests = [
    (("IVF-iDET-2α", "IVF-iDET-2β"), 'Mann-Whitney'),
    (("IVF-iDET-4α", "IVF-iDET-4β"), 'Mann-Whitney'),
]

# 分别执行统计检验并添加注释
for pair, test in pairs_with_tests:
    annotator = Annotator(
        ax,
        [pair],
        data=data,
        x="DNA-hAT-Charlie",
        y="CytoTRACE2_Score",
        order=ordered_phenotypes
    )
    annotator.configure(
        test=test,
        text_format='star',
        line_height=0.01,
        line_width=1,
        loc='outside',
        fontsize=20
    )
    annotator.apply_and_annotate()

# 替换横坐标标签
new_labels = [x_label_mapping[label] for label in ordered_phenotypes]
ax.set_xticklabels(new_labels, fontsize=20)

# 坐标轴格式设置
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.ylabel('CytoTRACE2 Score', fontsize=20)
plt.xlabel('')
plt.tight_layout()
plt.title('DNA-hAT-Charlie', fontsize=24, pad=50)

# 保存为 SVG 矢量图
plt.savefig(
    'IVF_CytoTRACE2_DNA-hAT-Charlie.svg',
    dpi=300,
    bbox_inches='tight',
    transparent=True
)
plt.show()
```
