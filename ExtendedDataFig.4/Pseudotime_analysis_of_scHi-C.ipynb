```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator

# =========================
# 读取数据
# =========================
data = pd.read_csv('IVF_TADcluster_time.txt', sep='\t')

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['axes.unicode_minus'] = False

sns.set(style="white")  # Nature-style 白底

# =========================
# ID 替换：旧标签 -> 新标签
# =========================
id_mapping = {
    "IVF-2C-C1": "IVF-TLD-2α",
    "IVF-2C-C2": "IVF-TLD-2β",
    "IVF-4C-D1": "IVF-TLD-4α",
    "IVF-4C-D2": "IVF-TLD-4β",
    "UN": "UN",
}

data['type'] = data['type'].replace(id_mapping)

# =========================
# 固定顺序
# =========================
full_order = [
    "IVF-TLD-2α",
    "IVF-TLD-2β",
    "IVF-TLD-4α",
    "IVF-TLD-4β",
    "UN",
]

# 只保留存在的数据类别
order = [x for x in full_order if x in data["type"].unique()]

# =========================
# 调色板（沿用 color_mapping_raw）
# =========================
color_mapping = {
    "IVF-TLD-2α": "#fb9795",
    "IVF-TLD-2β": "#E9B44C",
    "IVF-TLD-4α": "#7c5e8c",
    "IVF-TLD-4β": "#b0e7ea",
    "UN": "lightgrey",
}

palette = [color_mapping.get(x, "grey") for x in order]

# =========================
# 计算均值
# =========================
means = data.groupby("type")["time"].mean().reindex(order)

# =========================
# 绘图
# =========================
fig, ax = plt.subplots(figsize=(10, 6))

sns.violinplot(
    x="type",
    y="time",
    data=data,
    order=order,
    palette=color_mapping,
    inner=None,
    cut=0,
    scale="width",
    ax=ax
)

sns.stripplot(
    x="type",
    y="time",
    data=data,
    order=order,
    color="black",
    size=3,
    alpha=0.5,
    jitter=True,
    ax=ax
)

# =========================
# 均值线 + 数值
# =========================
for i, g in enumerate(order):
    if g not in means or pd.isna(means[g]):
        continue
    m = means[g]
    ax.hlines(y=m, xmin=i-0.25, xmax=i+0.25, color="red", linewidth=3, zorder=5)
    ax.text(i+0.35, m, f"{m:.3f}", color="red", fontsize=12, va="center")

# =========================
# 统计检验：使用新标签
# =========================
pairs = [
    ("IVF-TLD-2α", "IVF-TLD-2β"),
    ("IVF-TLD-4α", "IVF-TLD-4β"),
]
pairs = [p for p in pairs if all(x in order for x in p)]

annotator = Annotator(
    ax,
    pairs,
    data=data,
    x="type",
    y="time",
    order=order
)

annotator.configure(
    test="Mann-Whitney",
    text_format="star",
    loc="inside",
    fontsize=16
)

annotator.apply_and_annotate()

# =========================
# 横坐标标签映射：直接显示新标签
# =========================
label_mapping = {
    "IVF-TLD-2α": "IVF-TLD-2α",
    "IVF-TLD-2β": "IVF-TLD-2β",
    "IVF-TLD-4α": "IVF-TLD-4α",
    "IVF-TLD-4β": "IVF-TLD-4β",
    "UN": "UN",
}

ax.set_xticklabels([label_mapping[x] for x in order], fontsize=16)

# =========================
# 风格
# =========================
ax.set_ylabel("Pseudotime", fontsize=18)
ax.set_xlabel("")

ax.grid(False)
sns.despine()
plt.tight_layout()

# =========================
# 保存
# =========================
plt.savefig("IVF_pseudotime_styled.svg", dpi=300, bbox_inches="tight", transparent=True)
plt.show()
```

```python
import scanpy as sc
import anndata as ad
from py_monocle import learn_graph, order_cells, compute_cell_states

# 创建 dummy 表达矩阵，只需要 UMAP 坐标即可
adata = ad.AnnData(np.zeros((y.shape[0], 0)))
adata.obsm['X_umap'] = y

# 计算邻接图
sc.pp.neighbors(adata, use_rep='X_umap', n_neighbors=15)

# 使用 Leiden 聚类（更适合 Monocle 风格的图结构学习）
sc.tl.leiden(adata, resolution=0.1)
leiden_clusters = adata.obs['leiden'].astype(int).values

projected_points, mst, centroids = learn_graph(matrix=y, clusters=leiden_clusters)

pseudotime = order_cells(
    y,
    centroids,
    mst=mst,
    projected_points=projected_points,
    root_cells=0  # 可以选择起始细胞，比如胚胎早期的
)
```

```python
import matplotlib.pyplot as plt
import numpy as np

edges = np.array(mst.nonzero()).T
ds = 10

# 主图结构
plt.figure(figsize=(6, 4))
plt.title("Principal Graph")
plt.scatter(y[:, 0], y[:, 1], c=leiden_clusters, s=ds, cmap="tab10", edgecolors='none', alpha=0.8, rasterized=True)
for edge in edges:
    plt.plot(centroids[edge, 0], centroids[edge, 1], c="black", linewidth=1)
plt.xticks([]); plt.yticks([])
plt.savefig('PA_Principal_Graph.svg', dpi=300, bbox_inches='tight', transparent=True)
plt.show()

# 细胞状态图
cell_states, _ = compute_cell_states(y, centroids, mst)
plt.figure(figsize=(6, 4))
plt.title("Cell States")
plt.scatter(y[:, 0], y[:, 1], c=cell_states, s=ds, cmap="nipy_spectral", edgecolors='none', alpha=0.8, rasterized=True)
for edge in edges:
    plt.plot(centroids[edge, 0], centroids[edge, 1], c="black", linewidth=1)
plt.colorbar()
plt.xticks([]); plt.yticks([])
plt.savefig('PA_Cell_States.svg', dpi=300, bbox_inches='tight', transparent=True)
plt.show()

# 拟时序图
plt.figure(figsize=(6, 4))
plt.title("Pseudotime")
plt.scatter(y[:, 0], y[:, 1], c=pseudotime, s=ds, cmap="plasma", edgecolors='none', alpha=0.8, rasterized=True)
plt.colorbar()
plt.xticks([]); plt.yticks([])
plt.savefig('PA_Pseudotime.svg', dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator

# =========================
# 读取数据
# =========================
data = pd.read_csv('PA_TADcluster_time.txt', sep='\t')

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['axes.unicode_minus'] = False

sns.set(style="white")  # Nature-style 白底

# =========================
# 固定顺序 & 调色板
# =========================
full_order = ["PA-2C-A1", "PA-2C-A2", "PA-4C-B1", "PA-4C-B2", "UN"]
order = [x for x in full_order if x in data["type"].unique()]

color_mapping = {
    "PA-2C-A1": "#7dace4",
    "PA-2C-A2": "#ff165d",
    "PA-4C-B1": "green",
    "PA-4C-B2": "orange",
    "UN": "lightgrey"
}

palette = [color_mapping.get(x, "grey") for x in order]

# =========================
# 均值计算
# =========================
means = data.groupby("type")["time"].mean().reindex(order)

# =========================
# 绘图
# =========================
fig, ax = plt.subplots(figsize=(10, 6))

sns.violinplot(
    x="type",
    y="time",
    data=data,
    order=order,
    palette=color_mapping,
    inner=None,
    cut=0,
    scale="width",
    ax=ax
)

sns.stripplot(
    x="type",
    y="time",
    data=data,
    order=order,
    color="black",
    size=3,
    alpha=0.5,
    jitter=True,
    ax=ax
)

# =========================
# 均值线 + 数值显示
# =========================
for i, g in enumerate(order):
    if g not in means or pd.isna(means[g]):
        continue
    m = means[g]
    ax.hlines(y=m, xmin=i-0.25, xmax=i+0.25, color="red", linewidth=3, zorder=5)
    ax.text(i+0.4, m, f"{m:.3f}", color="red", fontsize=12, va="center")

# =========================
# 统计检验
# =========================
pairs = [("PA-2C-A1", "PA-2C-A2"), ("PA-4C-B1", "PA-4C-B2")]
pairs = [p for p in pairs if all(x in order for x in p)]  # 过滤不存在的类别

annotator = Annotator(
    ax,
    pairs,
    data=data,
    x="type",
    y="time",
    order=order
)
annotator.configure(
    test="Mann-Whitney",
    text_format="star",
    loc="inside",
    fontsize=16
)
annotator.apply_and_annotate()

# =========================
# 横坐标标签映射
# =========================
label_mapping = {
    "PA-2C-A1": "PA-2C-H1",
    "PA-2C-A2": "PA-2C-H2",
    "PA-4C-B1": "PA-4C-H1",
    "PA-4C-B2": "PA-4C-H2",
    "UN": "UN"
}
ax.set_xticklabels([label_mapping[x] for x in order], fontsize=16)

# =========================
# 风格美化
# =========================
ax.set_ylabel("Pseudotime", fontsize=18)
ax.set_xlabel("")
#ax.set_title("PA developmental pseudotime", fontsize=20)

ax.grid(False)
sns.despine()
plt.tight_layout()

# =========================
# 保存
# =========================
plt.savefig("PA_pseudotime_styled.svg", dpi=300, bbox_inches="tight", transparent=True)
plt.show()
```
