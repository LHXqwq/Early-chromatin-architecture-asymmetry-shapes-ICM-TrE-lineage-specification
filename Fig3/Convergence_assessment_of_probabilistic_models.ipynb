```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['svg.hashsalt'] = 'hello'

plt.rcParams.update({
    'font.size': 14,
    'axes.titlesize': 18,
    'axes.labelsize': 16,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 11
})

data = pd.read_csv("IVF_150p.txt", sep="\t", header=None).values
cum_mean = np.cumsum(data, axis=0) / np.arange(1, data.shape[0] + 1)[:, None]

plt.figure(figsize=(9, 7))

# ---------- 4 种颜色，每组 10 条同色 ----------
bases = ['#fb9795', '#E9B44C', 'lightgrey', '#A5D6A7']
colors = [bases[i // 10] for i in range(40)]   # 前10→#fb9795, 11-20→#E9B44C, ...
# ----------------------------------------

for i in range(40):
    plt.plot(cum_mean[:, i], alpha=0.75, color=colors[i], linewidth=1.2)

plt.title("IVF GA Simulation: Convergence of Probability Estimates", pad=15)
plt.xlabel("Simulation Number")
plt.ylabel("Cumulative Mean")

# ---------- 只显示 4 个图例条目 ----------
legend_handles = [Line2D([0], [0], color=b, lw=2.0, alpha=0.9) for b in bases]
legend_labels = [
    r'$p^{1}_{1}$–$p^{1}_{10}$',
    r'$p^{2}_{1}$–$p^{2}_{10}$',
    r'$p^{3}_{1}$–$p^{3}_{10}$',
    r'$p^{4}_{1}$–$p^{4}_{10}$',
]
plt.legend(
    legend_handles, legend_labels,
    title='Groups',
    bbox_to_anchor=(1.02, 1),
    loc='upper left',
    borderaxespad=0,
    frameon=False,
    fontsize=11,
    title_fontsize=12
)
# ----------------------------------------

plt.subplots_adjust(right=0.80)
plt.savefig('IVF_GA_Simulation150.svg', dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```
